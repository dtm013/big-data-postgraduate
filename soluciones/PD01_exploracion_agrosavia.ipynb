## 2. Diagnóstico de cinco campos · 10 minutos

Construir una tabla llamada `perfil` para `Secuencial`, `Fecha de Análisis`, `Departamento`, `Municipio` y `pH agua:suelo`. Para cada campo se registran los nulos de pandas, los textos vacíos después de retirar espacios y los marcadores `ND`. Las tres situaciones se cuentan por separado.

📘 **Apoyo:** `isna`, `.str.strip`, `.str.upper`, `.eq` y `sum`.

✅ **Comprobación:** `perfil` debe tener una fila por campo. Con `keep_default_na=False`, un texto vacío o `ND` no se convierte automáticamente en un nulo.

````python
# Desarrollo de la actividad 2
# Crear y mostrar perfil con las cinco columnas solicitadas.
# Desarrollo de la actividad 2
campos = ["Secuencial", "Fecha de Análisis", "Departamento", "Municipio", "pH agua:suelo"]
texto = suelo[campos]

perfil = pd.DataFrame({
    "nulos_pandas": texto.isna().sum(),
    "vacios_texto": texto.apply(lambda col: col.str.strip().eq("").sum()),
    "marcadores_ND": texto.apply(lambda col: col.str.strip().eq("ND").sum())
})

display(perfil)
````

**Interpretación del resultado:**

_Al leer el archivo con keep_default_na=False, los nulos automáticos de pandas son cero (o casi cero). La información ausente en la fuente original está catalogada realmente como cadenas de texto vacías ("") o bajo el marcador explícito "ND". Separar estos conteos nos da un diagnóstico real de la completitud de la base de datos sin alterar su estado original.._

## 3. Conversión de pH y fecha · 15 minutos

Crear `ph` desde `pH agua:suelo` y `fecha_dt` desde `Fecha de Análisis`. Utilizar conversión numérica con `errors="coerce"` y fecha con formato `%d/%m/%Y`. Conservar los originales. Distinguir entradas vacías de textos no vacíos que no pudieron convertirse.

📘 **Apoyo:** `pd.to_numeric`, `.str.replace`, `pd.to_datetime`, `isna` y máscaras booleanas.

✅ **Comprobación:** Se muestran los conteos de conversiones fallidas y hasta cinco ejemplos con su valor original. No se reemplazan medidas ausentes por cero.

````python
# Desarrollo de la actividad 3
# Crear suelo["ph"] y suelo["fecha_dt"].
# Mostrar los fallos de conversión y sus valores originales.
# Desarrollo de la actividad 3
# 1. Reemplazar coma por punto y convertir a numérico
suelo["ph"] = pd.to_numeric(
    suelo["pH agua:suelo"].str.strip().str.replace(",", ".", regex=False),
    errors="coerce"
)

# 2. Convertir a fecha con el formato día/mes/año
suelo["fecha_dt"] = pd.to_datetime(
    suelo["Fecha de Análisis"],
    format="%d/%m/%Y",
    errors="coerce"
)

# Identificar fallos
vacia_ph = suelo["pH agua:suelo"].str.strip().isin(["", "ND"])
no_numerica = ~vacia_ph & suelo["ph"].isna()

vacia_fecha = suelo["Fecha de Análisis"].str.strip().isin(["", "ND"])
no_fecha = ~vacia_fecha & suelo["fecha_dt"].isna()

print("Fallos en pH (textos anómalos):", no_numerica.sum())
display(suelo.loc[no_numerica, ["Secuencial", "pH agua:suelo"]].head())

print("\nFallos en Fecha (no interpretables):", no_fecha.sum())
display(suelo.loc[no_fecha, ["Secuencial", "Fecha de Análisis"]].head())
````

**Interpretación del resultado:**

_Se logró la conversión numérica (cambiando coma por punto) y la conversión de fechas. Al usar el parámetro errors="coerce", los valores que no cumplían el formato se transformaron en NaN o NaT. Es fundamental no usar fillna(0) para tapar estos huecos, ya que un pH de 0 distorsionaría toda la estadística agrícola; el dato simplemente se asume como ausente.._

## 4. Registros para revisión · 15 minutos

Crear la columna booleana `en_revision`. Se marca una fila si su pH es ausente, está fuera de `[0, 14]` o su fecha no se pudo interpretar. Contar por separado cada motivo y el total de filas marcadas. Conservar todas las observaciones.

📘 **Apoyo:** `between`, `isna`, `|`, `~`, `.loc` y `value_counts`.

✅ **Comprobación:** Las filas marcadas y no marcadas deben sumar el total original. Una misma fila puede tener varios motivos; por eso la suma de los motivos no necesariamente coincide con el total marcado.

````python
# Desarrollo de la actividad 4
# Crear en_revision y los conteos por motivo.
# Comprobar la conservación del total original con assert.
# Desarrollo de la actividad 4
suelo["en_revision"] = (
    suelo["ph"].isna() | 
    ~suelo["ph"].between(0, 14) | 
    suelo["fecha_dt"].isna()
)

motivo_ph_nulo = suelo["ph"].isna().sum()
motivo_ph_fuera = (~suelo["ph"].between(0, 14) & suelo["ph"].notna()).sum()
motivo_fecha_nula = suelo["fecha_dt"].isna().sum()

print("1. pH nulo:", motivo_ph_nulo)
print("2. pH fuera de [0, 14]:", motivo_ph_fuera)
print("3. Fecha nula/anómala:", motivo_fecha_nula)
print("Total de filas marcadas:", suelo["en_revision"].sum())

# Comprobación estricta de conservación
assert len(suelo) == total_filas_original, "Error: Se eliminaron filas."
print("¡El total de filas se conserva intacto!")
````

**Interpretación del resultado:**

_La etiqueta booleana en_revision nos permite identificar las filas anómalas sin tener que borrarlas de la base de datos (preservando el 100% de la tabla de AGROSAVIA). La suma individual de motivos supera el total de filas marcadas porque un mismo registro puede presentar múltiples errores a la vez (ej. no tener fecha y tampoco tener pH)._

## 5. Evidencias y conclusiones · 10 minutos

Construir `controles` con las columnas `control` y `cantidad`: total de entrada, pH sin número, pH fuera del intervalo, fecha no interpretable y total de filas en revisión. Exportar a `OUT / "controles_agrosavia.csv"` sin índice. Redactar dos hallazgos apoyados en los conteos.

📘 **Apoyo:** `pd.DataFrame`, `to_csv(index=False)` y las tablas de las actividades anteriores.

✅ **Comprobación:** El archivo se genera dentro de la ruta `OUT`. Cada hallazgo menciona un conteo y explica su significado o limitación.

````python
# Desarrollo de la actividad 5
# Construir controles, mostrarlo y exportarlo a OUT.
# Redactar los dos hallazgos en la celda Markdown siguiente.
# Desarrollo de la actividad 5
controles = pd.DataFrame({
    "control": [
        "Total de entrada", 
        "pH sin número", 
        "pH fuera del intervalo", 
        "Fecha no interpretable", 
        "Total de filas en revisión"
    ],
    "cantidad": [
        total_filas_original,
        motivo_ph_nulo,
        motivo_ph_fuera,
        motivo_fecha_nula,
        suelo["en_revision"].sum()
    ]
})

display(controles)

# Exportar el resumen a CSV
controles.to_csv(OUT / "controles_agrosavia.csv", index=False)
print("CSV generado correctamente en:", OUT)
````

**Interpretación del resultado:**

_Hallazgo 1 (Fechas nulas): Existe una porción de registros con fechas ausentes o inválidas. Aunque estas filas no sirven para hacer trazabilidad temporal o estacionalidad, no se deben eliminar porque sus otros campos son válidos para perfilamientos espaciales.
Hallazgo 2 (pH atípico): Se encontraron registros con mediciones de pH fuera del límite físico [0, 14]. Esto alerta sobre posibles errores de digitación o unidades desde el laboratorio. Haberlos etiquetado evita que sesguen los promedios departamentales sin necesidad de truncar la base de datos._
